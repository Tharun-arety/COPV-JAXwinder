# JAX COPV Optimization

This notebook runs the structured JAX optimization path from the package modules. It covers the three manufacturing
surrogates on the full COPV shell and writes the comparison outputs into `outputs/`.


## Setup


```python
import os

ENABLE_X64 = False
os.environ.setdefault("XLA_PYTHON_CLIENT_PREALLOCATE", "false")

from jax import config as jax_config
jax_config.update("jax_enable_x64", ENABLE_X64)

import json
import sys
import gc
from pathlib import Path

import jax
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Image, display


def find_project_root(start: Path) -> Path:
    start = start.resolve()
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").exists() and (candidate / "src").exists():
            return candidate
    raise FileNotFoundError("Could not locate the project root containing pyproject.toml and src/")


PROJECT_ROOT = find_project_root(Path.cwd())
SRC = PROJECT_ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

OUTPUTS = PROJECT_ROOT / "outputs"
OUTPUTS.mkdir(parents=True, exist_ok=True)


def hostify_tree(tree):
    return jax.tree_util.tree_map(
        lambda x: np.asarray(jax.device_get(x)) if isinstance(x, jax.Array) else x,
        tree,
    )

print(f"Project root : {PROJECT_ROOT}")
print(f"Outputs      : {OUTPUTS}")
print(f"JAX backend  : {jax.default_backend()}")
print(f"JAX devices  : {jax.devices()}")
print(f"JAX x64      : {ENABLE_X64}")

from copv_opt.config import GeometryConfig, IFPConfig, MaterialConfig, PatchConfig, WindingConfig
from copv_opt.geometry import copv_meridional_metrics, copv_surface_from_sphi_np, ensure_copv_mesh
from copv_opt.optimize import run_ifp_optimization, run_patch_optimization, run_winding_angle_sweep
from copv_opt.physics import baseline_response, build_copv_fem_state, make_solve_compliance
from copv_opt.visualize import (
    build_ifp_layout_data,
    build_patch_layout_data,
    build_winding_layout_data,
    compare_vtu_side_by_side,
    load_layout_json,
    plot_ifp_projection,
    plot_patch_projection,
    plot_winding_paths,
    render_explicit_manufacturing_layout,
    render_vtu_interactive,
    save_explicit_manufacturing_layout_screenshot,
    save_layout_json,
    save_tradeoff_plot,
    show_copv_mesh,
    write_vtu,
)
```

## COPV State Build


```python
PATCH_COUNT = 128
IFP_FAMILY_COUNT = 12

geom = GeometryConfig()
material = MaterialConfig()
patch_cfg = PatchConfig(count=PATCH_COUNT)
ifp_cfg = IFPConfig(family_count=IFP_FAMILY_COUNT)
winding_cfg = WindingConfig()

step_path = OUTPUTS / "copv_shell.step"
msh_path = OUTPUTS / "copv_shell.msh"
mesh = ensure_copv_mesh(step_path, msh_path, geom, remesh=not msh_path.exists())
state = build_copv_fem_state(mesh.nodes, mesh.elems, material, geom)
solve_compliance = make_solve_compliance(state)
dense_patch_tensor_bytes = patch_cfg.count * state["element_count"] * 81 * np.dtype(np.float64 if ENABLE_X64 else np.float32).itemsize
baseline = hostify_tree(baseline_response(state, material, solve_compliance))
gc.collect()

fig = show_copv_mesh(
    mesh.nodes,
    state["outer_faces"],
    geom,
    f"COPV analysis mesh: {len(mesh.nodes)} nodes / {len(mesh.elems)} tetra",
    OUTPUTS / "copv_analysis_mesh.png",
)
display(fig)
plt.close(fig)

print(f"Patch count            : {patch_cfg.count}")
print(f"IFP family count       : {ifp_cfg.family_count}")
print(f"Equivalent dense patch stiffness block : {dense_patch_tensor_bytes / 1024**2:.2f} MiB")
print("Patch stiffness accumulation uses a scan and does not materialize that dense block.")
print(f"Baseline strain energy : {float(np.asarray(baseline['compliance'])):.6f}")
```

## Patch and IFP L-BFGS Runs


```python
patch_run = hostify_tree(run_patch_optimization(state, material, patch_cfg, geom, solve_compliance))
jax.clear_caches()
gc.collect()

ifp_run = hostify_tree(run_ifp_optimization(state, material, ifp_cfg, geom, solve_compliance))
jax.clear_caches()
gc.collect()

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot([x["beta"] for x in patch_run["history"]], [x["strain_energy"] for x in patch_run["history"]], marker="o", color="steelblue")
ax.set_xlabel("beta")
ax.set_ylabel("strain energy")
ax.set_title("Patch L-BFGS continuation")
ax.grid(alpha=0.3)
fig.tight_layout()
display(fig)
fig.savefig(OUTPUTS / "patch_history.png", dpi=120)
plt.close(fig)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot([x["beta"] for x in ifp_run["history"]], [x["strain_energy"] for x in ifp_run["history"]], marker="o", color="darkorange")
ax.set_xlabel("beta")
ax.set_ylabel("strain energy")
ax.set_title("IFP L-BFGS continuation")
ax.grid(alpha=0.3)
fig.tight_layout()
display(fig)
fig.savefig(OUTPUTS / "ifp_history.png", dpi=120)
plt.close(fig)

print(json.dumps({"patch_history": patch_run["history"], "ifp_history": ifp_run["history"]}, indent=2))
```

## Winding Angle Sweep


```python
winding_run = hostify_tree(run_winding_angle_sweep(state, material, winding_cfg, geom, solve_compliance))
best_angle = winding_run["angle_deg"]

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot([x["angle_deg"] for x in winding_run["history"]], [x["strain_energy"] for x in winding_run["history"]], marker="o", color="forestgreen")
ax.set_xlabel("Cylinder angle [deg]")
ax.set_ylabel("strain energy")
ax.set_title("Winding angle sweep")
ax.grid(alpha=0.3)
fig.tight_layout()
display(fig)
fig.savefig(OUTPUTS / "winding_angle_sweep.png", dpi=120)
plt.close(fig)

winding_s = np.linspace(0.0, copv_meridional_metrics(geom.mid_radius, geom.cylinder_length, geom.opening_radius)[2], 220)
winding_rho = copv_surface_from_sphi_np(geom.mid_radius, winding_s, np.zeros_like(winding_s), geom.cylinder_length, geom.opening_radius)["rho"]
winding_alpha_profile = np.degrees(
    np.arcsin(
        np.clip(
            geom.mid_radius * np.sin(np.radians(best_angle)) / np.clip(winding_rho, geom.opening_radius + 1e-6, None),
            -0.98,
            0.98,
        )
    )
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(winding_s, winding_alpha_profile, color="seagreen")
ax.set_xlabel("Meridional surface coordinate")
ax.set_ylabel("Winding angle [deg]")
ax.set_title("Winding angle profile")
ax.grid(alpha=0.3)
fig.tight_layout()
display(fig)
fig.savefig(OUTPUTS / "winding_angle_profile.png", dpi=120)
plt.close(fig)

print(json.dumps({"winding_history": winding_run["history"], "best_angle_deg": best_angle}, indent=2))
```

## COPV Manufacturing Overlays


```python
fig = plot_patch_projection(patch_run["result"], patch_cfg, geom, OUTPUTS / "copv_patch_projection.png")
display(fig)
plt.close(fig)

fig = plot_ifp_projection(ifp_run["result"], ifp_cfg, geom, OUTPUTS / "copv_ifp_projection.png")
display(fig)
plt.close(fig)

fig, winding_paths = plot_winding_paths(best_angle, geom, winding_cfg, OUTPUTS / "copv_winding_paths.png")
display(fig)
plt.close(fig)
```

## VTU Export and Summary


```python
nodes = mesh.nodes
elems = mesh.elems

base_u = np.asarray(baseline["displacement"]).reshape(len(nodes), 3)
patch_u = np.asarray(patch_run["result"]["displacement"]).reshape(len(nodes), 3)
ifp_u = np.asarray(ifp_run["result"]["displacement"]).reshape(len(nodes), 3)
winding_u = np.asarray(winding_run["result"]["displacement"]).reshape(len(nodes), 3)

base_vtu = write_vtu(
    OUTPUTS / "copv_base.vtu",
    nodes,
    elems,
    base_u,
    np.asarray(baseline["thickness"]),
    np.asarray(baseline["density"]),
    np.asarray(baseline["fiber_dirs"]),
    np.asarray(baseline["coverage"]),
)
patch_vtu = write_vtu(
    OUTPUTS / "copv_patch_jax.vtu",
    nodes,
    elems,
    patch_u,
    np.asarray(patch_run["result"]["thickness"]),
    np.asarray(patch_run["result"]["density"]),
    np.asarray(patch_run["result"]["fiber_dirs"]),
    np.asarray(patch_run["result"]["coverage"]),
)
ifp_vtu = write_vtu(
    OUTPUTS / "copv_ifp_jax.vtu",
    nodes,
    elems,
    ifp_u,
    np.asarray(ifp_run["result"]["thickness"]),
    np.asarray(ifp_run["result"]["density"]),
    np.asarray(ifp_run["result"]["fiber_dirs"]),
    np.asarray(ifp_run["result"]["coverage"]),
)
winding_vtu = write_vtu(
    OUTPUTS / "copv_winding_jax.vtu",
    nodes,
    elems,
    winding_u,
    np.asarray(winding_run["result"]["thickness"]),
    np.asarray(winding_run["result"]["density"]),
    np.asarray(winding_run["result"]["fiber_dirs"]),
    np.asarray(winding_run["result"]["coverage"]),
)

patch_layout = build_patch_layout_data(patch_run["result"], patch_cfg, geom)
ifp_layout = build_ifp_layout_data(ifp_run["result"], ifp_cfg, geom)
winding_layout = build_winding_layout_data(best_angle, geom, winding_cfg)

patch_layout_path = save_layout_json(OUTPUTS / "patch_layout.json", patch_layout)
ifp_layout_path = save_layout_json(OUTPUTS / "ifp_layout.json", ifp_layout)
winding_layout_path = save_layout_json(OUTPUTS / "winding_layout.json", winding_layout)

pyvista_patch_explicit = save_explicit_manufacturing_layout_screenshot(
    base_vtu,
    OUTPUTS / "pyvista_patch_explicit_layout.png",
    patch_polygons=[entry["corners"] for entry in patch_layout["patches"]],
    patch_color="steelblue",
    patch_edge_color="midnightblue",
    title="Explicit Patch Layout over COPV",
)
pyvista_ifp_explicit = save_explicit_manufacturing_layout_screenshot(
    base_vtu,
    OUTPUTS / "pyvista_ifp_explicit_layout.png",
    curve_points_list=ifp_layout["curves"],
    curve_color="crimson",
    tow_radius=max(1.15, 0.12 * ifp_cfg.tow_width),
    title="Explicit IFP Layout over COPV",
)
pyvista_winding_explicit = save_explicit_manufacturing_layout_screenshot(
    base_vtu,
    OUTPUTS / "pyvista_winding_explicit_layout.png",
    curve_points_list=[entry["points"] for entry in winding_layout["paths"]],
    curve_colors=[
        "forestgreen" if entry["handedness"] == "clockwise" else "darkmagenta"
        for entry in winding_layout["paths"]
    ],
    tow_radius=1.15,
    title="Explicit Winding Layout over COPV",
)

display(Image(filename=str(pyvista_patch_explicit)))
display(Image(filename=str(pyvista_ifp_explicit)))
display(Image(filename=str(pyvista_winding_explicit)))

fig = save_tradeoff_plot(
    OUTPUTS / "strain_energy_vs_mass.png",
    labels=["Base", "Patch-JAX", "IFP-JAX", "Winding-JAX"],
    masses=[
        float(np.asarray(baseline["mass_metric"])),
        patch_run["history"][-1]["mass_metric"],
        ifp_run["history"][-1]["mass_metric"],
        winding_run["history"][np.argmin([x["strain_energy"] for x in winding_run["history"]])]["mass_metric"],
    ],
    strain_energies=[
        float(np.asarray(baseline["compliance"])),
        patch_run["history"][-1]["strain_energy"],
        ifp_run["history"][-1]["strain_energy"],
        min(x["strain_energy"] for x in winding_run["history"]),
    ],
)
display(fig)
plt.close(fig)

summary = {
    "jax_backend": jax.default_backend(),
    "mesh": {
        "nodes": int(len(nodes)),
        "elements": int(len(elems)),
        "step": str(step_path),
        "msh": str(msh_path),
        "mesh_hmin": float(geom.mesh_hmin),
        "mesh_hmax": float(geom.mesh_hmax),
    },
    "baseline": {
        "strain_energy": float(np.asarray(baseline["compliance"])),
        "mass_metric": float(np.asarray(baseline["mass_metric"])),
        "vtu": str(base_vtu),
    },
    "patch_jax": {
        "history": patch_run["history"],
        "strain_energy": patch_run["history"][-1]["strain_energy"],
        "mass_metric": patch_run["history"][-1]["mass_metric"],
        "improvement_pct": float(100.0 * (float(np.asarray(baseline["compliance"])) - patch_run["history"][-1]["strain_energy"]) / float(np.asarray(baseline["compliance"]))),
        "vtu": str(patch_vtu),
    },
    "ifp_jax": {
        "history": ifp_run["history"],
        "strain_energy": ifp_run["history"][-1]["strain_energy"],
        "mass_metric": ifp_run["history"][-1]["mass_metric"],
        "improvement_pct": float(100.0 * (float(np.asarray(baseline["compliance"])) - ifp_run["history"][-1]["strain_energy"]) / float(np.asarray(baseline["compliance"]))),
        "vtu": str(ifp_vtu),
    },
    "winding_jax": {
        "history": winding_run["history"],
        "strain_energy": min(x["strain_energy"] for x in winding_run["history"]),
        "mass_metric": winding_run["history"][np.argmin([x["strain_energy"] for x in winding_run["history"]])]["mass_metric"],
        "improvement_pct": float(100.0 * (float(np.asarray(baseline["compliance"])) - min(x["strain_energy"] for x in winding_run["history"])) / float(np.asarray(baseline["compliance"]))),
        "cylinder_angle_deg": float(best_angle),
        "vtu": str(winding_vtu),
    },
    "visualisations": {
        "patch_plot": str(OUTPUTS / "copv_patch_projection.png"),
        "ifp_plot": str(OUTPUTS / "copv_ifp_projection.png"),
        "winding_plot": str(OUTPUTS / "copv_winding_paths.png"),
        "patch_layout_json": str(patch_layout_path),
        "ifp_layout_json": str(ifp_layout_path),
        "winding_layout_json": str(winding_layout_path),
        "pyvista_patch_explicit": str(pyvista_patch_explicit),
        "pyvista_ifp_explicit": str(pyvista_ifp_explicit),
        "pyvista_winding_explicit": str(pyvista_winding_explicit),
        "winding_clairaut_radius": float(winding_paths["clairaut_radius"]),
        "winding_path_count": int(winding_paths["family_count"]),
    },
    "comparison_plot": str(OUTPUTS / "strain_energy_vs_mass.png"),
}

summary_path = OUTPUTS / "summary.json"
summary_path.write_text(json.dumps(summary, indent=2), encoding="utf-8")
print(json.dumps(summary, indent=2))
```
